# 2. Data Understanding & Preparation

Import Libraries

Dataset contains hotel names and customer reviews.

```
import pandas as pd
import numpy as np
import re
import nltk

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```
df = pd.read_csv('data/hotel_reviews.csv')
df.head()
```

```
df.info()

df.isnull().sum()
```

## Data Cleaning & Preprocessing

Steps performed:
- Removed missing or null reviews
- Converted text to lowercase
- Removed punctuation and special characters
- Removed stopwords
- Prepared clean text for vectorization

#Text Cleaning Function (Code)

```
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = text.split()
    tokens = [w for w in tokens if w not in stop_words]
    return ' '.join(tokens)

df['clean_review'] = df['review'].apply(clean_text)
df.head()
```

# 3. Model / System Design

## AI Technique Used
Natural Language Processing (NLP) with Machine Learning-based similarity matching.

## Architecture / Pipeline
1. Input customer reviews
2. Text preprocessing and cleaning
3. TF-IDF vectorization
4. Cosine similarity computation
5. Top-N hotel recommendations

## Justification of Design Choices
TF-IDF combined with cosine similarity is computationally efficient,
interpretable, and effective for text-based recommendation systems.
TF-IDF + Cosine Similarity based Recommendation System.

#  Core Implementation
📌 TF-IDF Vectorization

```
tfidf = TfidfVectorizer(max_features=5000)
tfidf_matrix = tfidf.fit_transform(df['clean_review'])
```

## Cosine Similarity Computation

📌 Cosine Similarity Computation

Cosine similarity measures the similarity between customer reviews.

```
similarity_matrix = cosine_similarity(tfidf_matrix)
```

# 4. Recommendation Function
This function recommends top hotels based on similarity to a selected hotel.


```
def recommend_hotels(hotel_name, top_n=5):
    if hotel_name not in df['hotel_name'].values:
        return 'Hotel not found.'
    idx = df[df['hotel_name']==hotel_name].index[0]
    scores = list(enumerate(similarity_matrix[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    return [df.iloc[i[0]]['hotel_name'] for i in scores[1:top_n+1]]
```

📌 Sample Recommendation Output

```
recommend_hotels("Hotel Sunshine", top_n=5)
```
